To purge the unfinished scan entries :

A scan that was interrupted leaves placeholder rows (`loss=null`, `time=0.0`) that later runs consider as computed. They are removed here so that `20_MNESIS_scanning-parameters.ipynb` recomputes them. This applies to the scans of all batches (every `datetag`) in `cached_data/`. The scans that hold a `.lock` file are skipped (`remove_locks=False`), as they may still be running in another process. Pass `remove_locks=True` to also delete the `.lock` files of interrupted scans, only when no other scan is running.

```python
from mnesis_boilerplate import purge_unfinished_scans
purge_unfinished_scans(remove_locks=False, dry_run=False)
```
