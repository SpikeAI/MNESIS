To purge the unfinished scan entries :

A scan that was interrupted leaves placeholder rows (`loss=null`, `time=0.0`) that later runs consider as computed. They are removed here so that `20_MNESIS_scanning-parameters.ipynb` recomputes them. This applies to the scans of all batches (every `datetag`) in `cached_data/`. Their `.lock` files are also deleted (`remove_locks=True`), as none can be legitimate when this orchestrator starts: do not run it while another scan is running.

```python
from mnesis_boilerplate import purge_unfinished_scans
purge_unfinished_scans(remove_locks=False, dry_run=False)
```
